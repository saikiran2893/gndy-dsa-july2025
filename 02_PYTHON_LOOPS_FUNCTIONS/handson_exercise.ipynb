Write a Python program that categorizes a student's grade based on their test score.


```python
def stundents_grade(score):
    if score > 90:
        print(" A Grade ")
    elif score > 70:
        print(" B Grade")
    elif score > 50:
        print("C Grade")
    elif score > 45:
        print( " D Grade")
    else:
        print( " Fail")

score = input()
print(score)
stundents_grade(int(score))
```

Create a Python program that checks if a given number is even and positive, even and negative, odd and positive, or odd and negative.


```python
def number_check(num):
    if num%2 ==0:
        if num >0 :
            print(" Positive and Even")
        else:
            print( " Negative and Even")

    else:
        if num >0 :
            print(" Positive and Odd")
        else:
            print( " Negative and Odd")


num1 = input()
print(num1)
number_check(int(num1))
```

Question 1 (for loop with strings):
Write a Python program that uses a for loop to reverse a given string and prints the reversed string.


```python
string = "Datascience"
temp_char=''
for char in string:
    #print(char)
    output_string = char + temp_char
    #print(output_string)
    temp_char = output_string

print(f"final reverse string is {output_string}")
```

Question 1: You are building a shopping cart application. How would you use functions to calculate the total price of items in a cart with different quantities and prices?


```python
cart = {'apple':[2,50],
        'Banana':[12,5],
        'Oranges':[6,20]}

def total_prince(cart):
    final_bill = 0
    for key, value in cart.items():
       product_items = key
       quantity = value[0]
       items = value[1]
       total_value = quantity * items
       print(product_items, total_value)
       final_bill = final_bill+total_value
    return final_bill

total_prince(cart)
```
