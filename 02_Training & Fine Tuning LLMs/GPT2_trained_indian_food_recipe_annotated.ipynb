# **Cleaning Indian Food Recipe Dataset**



```python
# Intermediate preprocessing or setup step
# Import libraries

import pandas as pd
import re
```

```python
# Loads dataset using read_csv
# Load the Indian Food dataset
data = pd.read_csv("/content/sample_data/IndianFoodDataset.csv")
```

```python
# Displays sample rows or data overview
data.head()
```

```python
# Intermediate operation in preprocessing
data.info()
```

```python
# Drops NaN values or irrelevant columns
# Drop specified columns
data = data.drop(['Srno', 'RecipeName', 'Ingredients', 'Instructions'], axis=1)
```

```python
# Prints or shows data head/info
data.head()
```

```python
# Intermediate preprocessing step
data.isnull().sum()
```

```python
# Data cleaning by dropping unnecessary data
# Drop rows with missing values
data = data.dropna()
```

```python
# Intermediate processing step
def clean_text(text):
    # Remove non-alphanumeric characters and extra whitespaces
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# Apply text cleaning to relevant columns
text_columns = ['TranslatedRecipeName', 'TranslatedIngredients', 'TranslatedInstructions']
for column in text_columns:
    data[column] = data[column].apply(clean_text)
```

```python
# Intermediate processing step
# Save the cleaned dataset
data.to_csv("/content/sample_data/cleaned_IndianFoodDataset.csv", index=False)
```

# **Training GPT2 Model on cleaned_IndianFoodDataset**

```python
# Initializes GPT2 tokenizer
import pandas as pd
from sklearn.model_selection import train_test_split
from transformers import GPT2LMHeadModel, GPT2Tokenizer, TextDataset, DataCollatorForLanguageModeling, Trainer, TrainingArguments
```

```python
# Reloads the dataset (possibly cleaned version)
# Load the cleaned dataset

data = pd.read_csv("/content/sample_data/cleaned_IndianFoodDataset.csv")
```

```python
# Intermediate step likely involving token formatting
# Split the dataset into train, validation, and test sets

train_data, test_data = train_test_split(data, test_size=0.2, random_state=42)
train_data, validation_data = train_test_split(train_data, test_size=0.1, random_state=42)
```

```python
# Possibly dataset splitting or formatting
# Save the split datasets

train_data.to_csv("/content/sample_data/train_dataset.csv", index=False)
validation_data.to_csv("/content/sample_data/validation_dataset.csv", index=False)
test_data.to_csv("/content/sample_data/test_dataset.csv", index=False)
```

```python
# GPT2 tokenizer setup for model training
# Load the GPT-2 tokenizer and model
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')
```

```python
# Prepares dataset/tokenization further
# Tokenize the datasets
train_dataset = TextDataset(
    tokenizer=tokenizer,
    file_path="/content/sample_data/train_dataset.csv",
    block_size=128
)

validation_dataset = TextDataset(
    tokenizer=tokenizer,
    file_path="/content/sample_data/validation_dataset.csv",
    block_size=128
)

test_dataset = TextDataset(
    tokenizer=tokenizer,
    file_path="/content/sample_data/test_dataset.csv",
    block_size=128
)
```

```python
train_dataset
```

```python
# Prepares dataset/tokenization further
# Set up training arguments
training_args = TrainingArguments(
    output_dir="/content/sample_data/gpt2-indian-food",
    overwrite_output_dir=True,
    num_train_epochs=3,
    per_device_train_batch_size=2,
    save_steps=10_000,
    save_total_limit=2,
    prediction_loss_only=True,
)
```

```python
# Configures and starts training using Hugging Face Trainer
# Set up trainer
trainer = Trainer(
    model=model,
    args=training_args,
    data_collator=DataCollatorForLanguageModeling(
        tokenizer=tokenizer,
        mlm=False
    ),
    train_dataset=train_dataset,
    eval_dataset=validation_dataset,
)
```

```python
# Intermediate model training or evaluation setup
# Start training

trainer.train()
```

```python
# Intermediate logic (could be checkpoint save, logging, etc)
# Save the model

trainer.save_model("/content/sample_data/gpt2-indian-food")

# Save the configuration file
model = trainer.model
model.config.save_pretrained("/content/sample_data/gpt2-indian-food_config.json")
```

```python
# Sample output or model predictions printed
# Evaluate on the test set

results = trainer.evaluate(test_dataset)
print(results)
```

# **Model Evalution - BLUE & ROUGE**

```python
# Evaluation setup or calculation logic
import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer

# Download NLTK resources
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
```

```python
# Prints results from evaluation
# Load the trained model and tokenizer
model = GPT2LMHeadModel.from_pretrained("/content/sample_data/gpt2-indian-food")
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

# Function to generate text using the model
def generate_text(prompt, max_length=100, temperature=1.0):
    input_ids = tokenizer.encode(prompt, return_tensors="pt")
    output = model.generate(input_ids, max_length=max_length, temperature=temperature, num_beams=5, no_repeat_ngram_size=2)
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    return generated_text

# Example prompts and references
prompts = ["To make Masala Karela,",
           "Tomato Puliogere is prepared by",
           "Ragi Vermicelli Recipe starts with",
           "Gongura Chicken Curry Recipe first prep all the ingredients",
           "Andhra Style Alam Pachadi is made by"]

references = ["To begin making the Masala Karela Recipe, de-seed the karela and slice.",
              "To make tomato puliogere, first cut the tomatoes.",
              "To begin making the Ragi Vermicelli Recipe, first steam the ragi vermicelli.",
              "To begin making Gongura Chicken Curry Recipe, first prep all the ingredients.",
              "To make Andhra Style Alam Pachadi, first heat oil in a pan."]

# Generate text and calculate BLEU and ROUGE scores
smooth = SmoothingFunction().method1  # Define smoothing function for BLEU
bleu_scores = []
rouge_scores = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

for prompt, reference in zip(prompts, references):
    generated_text = generate_text(prompt)

    # BLEU score
    bleu_score = sentence_bleu([reference.split()], generated_text.split(), smoothing_function=smooth)
    bleu_scores.append(bleu_score)
    print(f"Prompt: {prompt}\nGenerated: {generated_text}\nReference: {reference}\nBLEU Score: {bleu_score:.4f}\n")

    # ROUGE scores
    rouge_score = rouge_scores.score(generated_text, reference)
    print(f"ROUGE Scores: {rouge_score}\n")

# Average BLEU score
average_bleu = sum(bleu_scores) / len(bleu_scores)
print(f"Average BLEU Score: {average_bleu:.4f}")
```

```python
# Loads tokenizer for UI inference
import gradio as gr
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
# Reloads tokenizer again (possibly duplicate or different config)
# Load the GPT-2 tokenizer and model
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("/content/sample_data/gpt2-indian-food")  # Use the path where your model is saved
```

```python
# Generates predictions using the trained GPT2 model
def generate_text(prompt):
    input_ids = tokenizer.encode(prompt, return_tensors="pt")
    output = model.generate(input_ids, max_length=50, num_beams=5, no_repeat_ngram_size=2, top_k=50, top_p=0.95, temperature=0.7)
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    return generated_text
```

```python
# Another generation example or method
# Create a Gradio Interface
iface = gr.Interface(
    fn=generate_text,
    inputs="text",
    outputs="text",
    live=True,
    examples=[["To make a delicious curry, ", "you need to start with fresh vegetables and aromatic spices."]]
)
```

```python
# Prepares Gradio UI (could involve interface definition)
iface.launch()
```
