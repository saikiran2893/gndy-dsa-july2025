```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, SGDRegressor
```

```python

# 1. Generate synthetic data (house size vs price)
np.random.seed(42)
X = 2 * np.random.rand(100, 1)  # house size (100 samples)
y = 4 + 3 * X + np.random.randn(100, 1)  # price with some noise
```

```python
# 2. Fit with OLS (closed form)
ols_model = LinearRegression()
ols_model.fit(X, y)
y_pred_ols = ols_model.predict(X)
```

```python

# 3. Fit with Gradient Descent
sgd_model = SGDRegressor(max_iter=1000, tol=1e-3, eta0=0.1, random_state=42)
sgd_model.fit(X, y.ravel())  # y must be 1D
y_pred_sgd = sgd_model.predict(X)
```

```python
# 4. Compare coefficients
print("OLS coefficients: Intercept =", ols_model.intercept_[0], 
      "Slope =", ols_model.coef_[0][0])
print("SGD coefficients: Intercept =", sgd_model.intercept_[0], 
      "Slope =", sgd_model.coef_[0])
```

```python



# 5. Plot
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color="blue", label="Data points")
plt.plot(X, y_pred_ols, color="red", linewidth=2, label="OLS Line (Exact)")
plt.plot(X, y_pred_sgd, color="green", linestyle="--", linewidth=2, label="SGD Line (Approx)")
plt.xlabel("House Size (1000 sq ft)")
plt.ylabel("House Price ($1000s)")
plt.legend()
plt.title("OLS vs Gradient Descent in Simple Linear Regression")
plt.show()
```
