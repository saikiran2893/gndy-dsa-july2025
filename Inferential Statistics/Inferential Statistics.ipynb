```python
import numpy as np
from scipy.stats import ttest_1samp

# Data: wait times in minutes
wait_times = [
    4.2, 3.8, 3.5, 4.0, 4.5, 3.3, 4.2, 4.1, 3.9, 4.3, 3.7, 4.0, 4.2, 3.8, 3.6, 
    4.1, 3.9, 4.4, 4.0, 4.2, 3.7, 3.9, 4.0, 4.1, 4.3, 3.6, 3.9, 4.0, 4.2, 4.2,
]

# Population mean to test against
population_mean = 3  # Null hypothesis: μ = 3

# Perform one-sample t-test
t_stat, p_value = ttest_1samp(wait_times, population_mean)

# Compute additional statistics
sample_mean = np.mean(wait_times)
sample_std = np.std(wait_times, ddof=1)  # Use ddof=1 for sample standard deviation
sample_size = len(wait_times)

# Print t-test statistics
print(f"Sample Mean: {sample_mean:.3f}")
print(f"Sample Standard Deviation: {sample_std:.3f}")
print(f"Sample Size: {sample_size}")
print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_value:.3e}")
```

```python
## Z- Test
```

```python


import numpy as np
from scipy.stats import ttest_ind

# Test scores for Method A and Method B
group_a = np.array([85, 88, 92, 78, 90])
group_b = np.array([91, 89, 82, 87, 88])

# Perform the independent t-test
t_stat, p_value = ttest_ind(group_a, group_b)

# Output the results
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")



# Interpret the result
alpha = 0.05, 
#CI - 95%
if p_value < alpha:
    print("Reject the null hypothesis: There is no significant difference between the two groups.")
else:
    print("Fail to reject the null hypothesis: There is significant difference between the two groups.")
```

```python
##Let's perform a one-sample Z-test using the generated dataset
## to evaluate the effectiveness of the weight loss program.
```

```python
import random
from scipy.stats import norm

# Generate sample weight loss data
random.seed(42)  # For reproducibility
weight_loss_data = [random.uniform(0.5, 10) for _ in range(100)]
weight_loss_data
```

```python


# Known population standard deviation
population_std = 2.5  # lbs

# Hypothesized mean (no weight loss)
hypothesized_mean = 0

# Sample statistics
sample_mean = sum(weight_loss_data) / len(weight_loss_data)  # Mean
print(sample_mean)
sample_size = len(weight_loss_data)

# Calculate Standard Error
standard_error = population_std / (sample_size**0.5)

# Calculate Z-Statistic
z_score = (sample_mean - hypothesized_mean) / standard_error
z_score
```

```python


# Calculate p-value (two-tailed test)
p_value = 2 * (1 - norm.cdf(abs(z_score)))
p_value
```

```python


# Critical Z-value for alpha = 0.05 (two-tailed)
z_critical = norm.ppf(1 - 0.05 / 2)
z_critical
```

```python


# Print Results
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Z-Score: {z_score:.2f}")
print(f"P-Value: {p_value:.4f}")
print(f"Z-Critical (5% significance level): ±{z_critical:.2f}")

# Interpretation
if abs(z_score) > z_critical:
    print("Reject the null hypothesis: No Significant weight loss observed.")
else:
    print("Fail to reject the null hypothesis: There is significant weight loss observed.")
```

```python
## Chisquare test
## Testing the Association Between Gender and Voting Preferences  
```

```python
import numpy as np
from scipy.stats import chi2_contingency

# Contingency table
data = np.array([
    [30, 25, 20],  # Male
    [35, 30, 40]   # Female
])
data
```

```python


# Perform Chi-Square Test
chi2, p, dof, expected = chi2_contingency(data)

# Results
print(f"Chi-Square Statistic: {chi2:.2f}")
print(f"P-Value: {p:.4f}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:")
print(expected)

## Ho - No association between gender and voting prefernce.
## Ha -  There is an association between gender and voting prefernce.

# Interpretation
alpha = 0.05
if p < alpha:
    print("Reject the null hypothesis: There is no association between gender and voting preferences.")
else:
    print("Fail to reject the null hypothesis: There is a association between gender and voting preferences.")
```

```python
## Anova Test
```

```python
import scipy.stats as stats

# Data
school_1 = [85, 88, 90, 92, 86]
school_2 = [78, 82, 80, 85, 88]
school_3 = [92, 94, 89, 88, 90]

# Perform ANOVA
f_stat, p_value = stats.f_oneway(school_1, school_2, school_3)

# Results
print(f"F-Statistic: {f_stat:.2f}")
print(f"P-Value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant difference between school means.")
else:
    print("Fail to reject the null hypothesis: There is  no significant difference between school means.")
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# --- Post-hoc analysis (only if ANOVA is significant) ---
import pandas as pd

# Combine data into a DataFrame
data = pd.DataFrame({
    'score': school_1 + school_2 + school_3,
    'school': ['School 1']*len(school_1) + ['School 2']*len(school_2) + ['School 3']*len(school_3)
})

# Tukey’s HSD test
tukey = pairwise_tukeyhsd(endog=data['score'], groups=data['school'], alpha=0.05)
print(tukey)
```
